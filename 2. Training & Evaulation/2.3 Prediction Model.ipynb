# Random Forest Prediction

import dependnces!

```python
import pandas as pd
import numpy as np
import pickle
from sklearn.preprocessing import StandardScaler, LabelEncoder
```

unpickle model and encoder and scaler

```python
with open('random_forest_model.pkl', 'rb') as f:
    rf_model = pickle.load(f)
with open('scaler.pkl', 'rb') as f:
    scaler = pickle.load(f)
with open('label_encoder.pkl', 'rb') as f:
    label_encoder = pickle.load(f)
with open('X_columns.pkl', 'rb') as f:
    X_columns = pickle.load(f)
```

compile a final prediction function utilising standardscalaer, encoder, and the training data.

```python
def predict_round_winner(model, scaler, label_encoder, X_columns):
    print("=== round prediction ===")
    map_name = input("Map name (e.g., de_dust2): ")
    ct_money = float(input("total CT money: "))
    t_money = float(input("total T money: "))
    ct_players_alive = int(input("CT players alive: "))
    t_players_alive = int(input("T players alive: "))
    bomb_planted = int(input("Is bomb planted? (0 = No, 1 = Yes): "))

#map encoding into numeric value
    if map_name in label_encoder.classes_:
        map_encoded = label_encoder.transform([map_name])[0]
    else:
        print("Map not found in dataset. Try again.")
        return

#data dictioanry    
    input_dict = {
        "map": map_encoded,
        "ct_money": ct_money,
        "t_money": t_money,
        "ct_players_alive": ct_players_alive,
        "t_players_alive": t_players_alive,
        "bomb_planted": bomb_planted
    }

    input_df = pd.DataFrame([input_dict])
    input_df = input_df.reindex(columns=X_columns, fill_value=0)
    input_scaled = scaler.transform(input_df)
    prediction = model.predict(input_df)

    result = "Counter-terrorists" if prediction[0] == 0 else "Terrorists"
    print(f" Predicted Winner: {result}")
```

run the model!

```python
predict_round_winner(rf_model, scaler, label_encoder, X_columns)
```
